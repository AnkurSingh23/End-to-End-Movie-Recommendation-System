```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
movies = pd.read_csv('../data/movies.csv')
credits = pd.read_csv('../data/credits.csv')
movies.head()
```

```python
credits.head()
```

```python
credits.shape, movies.shape
```

```python
credits.rename(columns={'id': 'movie_id'}, inplace=True)

movies = movies.merge(credits, on='movie_id')

movies.head()
```

```python
movies.shape
```

```python
movies.columns
```

```python
movies = movies[['movie_id', 'title_x', 'overview', 'genres', 'keywords', 
                 'cast', 'crew', 'original_language', 'vote_average', 'popularity']]
movies.rename(columns={'title_x': 'title'}, inplace=True)
```

```python
movies.head()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.isnull().sum()
```

```python
import ast
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies[['genres', 'keywords']].head()
```

```python
def convert_cast(obj):
    L= []
    counter = 0
    for i in ast.literal_eval(obj):
        L.append(i['name'])
        counter+=1
        if counter == 3:
            break
    return L
movies['cast'] = movies['cast'].apply(convert_cast)
movies['cast'].head()
```

```python
def convert_crew(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
movies['crew'] = movies['crew'].apply(convert_crew)
movies['crew'].head()
```

```python
movies.rename(columns={'original_language': 'language','crew':'director'}, inplace=True)
movies.columns
```

```python
movies.head()
```

```python
movies['genres'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['director'] = movies['director'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies[['cast', 'director']].head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['director']
movies['tags'] = movies['tags'].apply(lambda x: " ".join(x))
movies['tags'] = movies['tags'].apply(lambda x: x.lower())
movies[['title', 'tags']].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(movies['tags']).toarray()
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
similarity.shape
```

```python
def recommend(movie):
    movie_index = movies[movies['title'] == movie].index[0]
    distances = similarity[movie_index]
    
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    for i in movies_list:
        print(movies.iloc[i[0]].title)
recommend('Avatar')
```

```python
import pickle

pickle.dump(movies, open('movies.pkl', 'wb'))
pickle.dump(similarity, open('similarity.pkl', 'wb'))
```

```python
from sklearn.neighbors import NearestNeighbors

knn = NearestNeighbors(metric='cosine', algorithm='brute')
knn.fit(vectors)
```

```python
def recommend_knn(movie):
    movie_index = movies[movies['title'] == movie].index[0]
    distances, indices = knn.kneighbors([vectors[movie_index]], n_neighbors=6)

    for i in indices[0][1:]:
        print(movies.iloc[i].title)
recommend_knn('Avatar')
```
